```python
import geopandas as gpd

boundary = gpd.read_file("../data/gis/pune_boundary.geojson")
roads = gpd.read_file("../data/gis/roads_raw.geojson")
bridges = gpd.read_file("../data/gis/bridges_raw.geojson")

roads = roads.to_crs(boundary.crs)
bridges = bridges.to_crs(boundary.crs)

roads_clipped = gpd.clip(roads, boundary)
bridges_clipped = gpd.clip(bridges, boundary)

print(len(roads_clipped), len(bridges_clipped))

```

```python
roads_clipped["road_id"] = [
    f"road_{i}" for i in range(len(roads_clipped))
]

bridges_clipped["bridge_id"] = [
    f"bridge_{i}" for i in range(len(bridges_clipped))
]


```

```python
roads_final = roads_clipped[["road_id", "highway", "geometry"]]
bridges_final = bridges_clipped[["bridge_id", "geometry"]]
```

```python
roads_final.to_file(
    "../data/gis/roads_final.geojson",
    driver="GeoJSON"
)

bridges_final.to_file(
    "../data/gis/bridges_final.geojson",
    driver="GeoJSON"
)
```
